# **Publishing and Sharing in Tableau  **


Question 1:
What is the key difference between publishing a dashboard on Tableau Public versus Tableau Server/Cloud in terms of accessibility and data security?
    Tableau Public
    Dashboards are publicly accessible to anyone on the internet.
    No strong data security or user-level access control.
    Best for portfolios, learning, and non-sensitive data.
    Tableau Server/Cloud
    Dashboards are shared securely within organizations.
    Supports authentication, permissions, and role-based access.
    Suitable for confidential business data.
    Key Difference
    Tableau Public	Tableau Server/Cloud
    Public access	Controlled/private access
    No strong security	Advanced security & permissions
    Open sharing	Enterprise collaboration


Question 2:
When publishing a Tableau dashboard, what happens to a live connection compared to an extract? How does this affect data refresh and performance?
    Live Connection
    Dashboard remains directly connected to the original database.
    Data updates in real time.
    Performance depends on database speed and network connectivity.
    Extract Connection
    Tableau creates a snapshot copy of data.
    Faster dashboard performance.
    Requires scheduled refreshes to keep data updated.
    Impact Comparison
    Feature	Live Connection	Extract
    Data Freshness	Real-time	Scheduled refresh
    Performance	Slower sometimes	Faster
    Database Dependency	High	Low

    Permissions and User Roles

    User Groups
    Roles
    Permissions

    View access
    Editing rights
    Download permissions
    Interaction capabilities

    in Tableau Server and Tableau Cloud.


Question 4:
If you want users to view a dashboard but not download the underlying data, which permission setting would you adjust in Tableau Server?

    You should disable:

    “Download Data” Permission

    This prevents users from:

    Downloading underlying datasets
    Exporting raw data

    while still allowing dashboard viewing and interaction.



Question 5:
Name two methods by which a Tableau dashboard can be embedded into a web page or website.
    Method 1: Embed Code (iFrame)
    Tableau provides HTML embed code.
    Can be added directly to websites.
    Method 2: JavaScript API
    Use Tableau JavaScript API for interactive embedding.
    Allows customization and dynamic control.



Question 6:
List two best practices you should follow when preparing a dashboard for public sharing to ensure data security and privacy.
    Best Practice 1

    Remove sensitive/confidential information such as:

    Personal data
    Financial details
    Customer identities
    Best Practice 2

    Use aggregated or anonymized data instead of raw records.

    Additional practices:

    Verify permissions
    Disable unnecessary downloads
    Test public visibility before publishing



Question 7:
After publishing a dashboard, users report a “Data not found” error. What is the most common cause related to data source settings?
    Most Common Cause

    Broken or inaccessible data source connection.

    This usually happens because:

    Database path changed
    Credentials expired
    Server/database unavailable
    File location changed after publishing


Question 8:
When embedding a Tableau dashboard into a secure internal portal, what two additional security measures should be implemented to ensure only authorized users can access the embedded content?
    Security Measure 1: User Authentication

    Implement:

    Single Sign-On (SSO)
    Login authentication
    Active Directory integration
    Security Measure 2: Row-Level or Role-Based Security

    Restrict dashboard access based on:

    User roles
    Departments
    Permission groups

    This ensures users only see authorized content.
